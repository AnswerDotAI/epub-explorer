# epub-explorer

> EPUB exploration skill for reading `.epub` books directly from the ZIP archive without extraction.

`Epx` reads epub files directly from ZIP, with no extraction needed. It auto-discovers a single `.epub` in the current directory, or you can pass a path explicitly.

### Installation

Install latest from the GitHub [repository][repo]:

```sh
$ pip install git+https://github.com/kafkasl/epub-explorer.git
```

[repo]: https://github.com/kafkasl/epub-explorer

## How to use

Use the `Epx()` class to load a book. By default it will try to grab one in the current folder (if it's the only one).

You can also pass a path to the book or set the env var `EPX_BOOK`.

````
b = Epx() # loads epub in the current folder
b
````

````
b = Epx(path='chuang-tzu.epub')
b
````

### Table of contents

`toc()` maps spine indices to their TOC labels — these indices are what you pass to `section()`:

````
b = Epx()
print(b.toc()[:300])
````

### Reading a section

`section()` converts any spine section to markdown:

````
print(b.section(5)[:200])
````

### Images

Images can be extracted by their epub-internal path:

````
b.images()
````

````
b.img('2384639977200309426_cover.jpg')
````
